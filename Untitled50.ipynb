1. Reading the data

```python
def getTextLinesFromFile(fileName):
  ret = []
  with open(fileName, "r") as file:
    lines = file.readlines()
    for line in lines:
      ret.append(line.strip())
    return ret
```

```python
lines = getTextLinesFromFile("dataset.txt")
print(len(lines))
```

2. Substitute course codes.

```python
substitutions = dict()
substitutes = getTextLinesFromFile("substitutions.txt")
for line in substitutes:
  oldCC, newCC = line.split("\t")
  substitutions[oldCC.upper()] = newCC

def getSubstituteCode(oldCC):
  return substitutions[oldCC.upper()]
```

```python
for i in range(1,len(lines)):
  print(lines[i].split("\t")[2])
  oldCC = lines[i].split("\t")[2]
  newCC = getSubstituteCode(oldCC)
  lines[i] = lines[i].replace(oldCC, newCC)
  print(lines[i].split("\t")[2])
  #print(getSubstituteCode(lines[i].split("\t")[0])
```

3. Create a list of course codes registered for each roll number. Use a dictionary for representing the roll numbers. The key is the roll number, the value is the list of course codes associated with the roll number.

```python
registrations = dict()
for i in range(1, len(lines)):
  entries = lines[i].split("\t")
  rollNo = entries[0]
  courseCode = entries[2]
  if rollNo in registrations:
    registrations[rollNo].append(courseCode)
  else:
    registrations[rollNo] = [courseCode]
print(len(registrations))
```

```python
for rollNo in registrations:
  print(rollNo, registrations[rollNo])
```

```python
from collections import Counter

def getDataFromRegistrations(registrations):
  ret = []
  for rollNo in registrations:
    ret.append(registrations[rollNo])
  return ret


def count_distinct_lists(list_of_lists):
    """Count occurrences of distinct lists, treating unsorted lists as equal"""
    # Convert each list to a sorted tuple for comparison
    sorted_tuples = [tuple(sorted(lst)) for lst in list_of_lists]
    return Counter(sorted_tuples)

# Example usage
data = getDataFromRegistrations(registrations)

result = count_distinct_lists(data)
print(result)
```
