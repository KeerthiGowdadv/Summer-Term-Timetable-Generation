1. Reading the data

```python
def getTextLinesFromFile(fileName):
  ret = []
  with open(fileName, "r") as file:
    lines = file.readlines()
    for line in lines:
      ret.append(line.strip())
    return ret
```

```python
lines = getTextLinesFromFile("dataset.txt")
print(len(lines))
```

```python
MOOC_COURSES = {
    "BBB2016",
    "CSE3111",
    "CSE3202",
    "MGT3001",
    "ENG3004",
    "MEC3301",
    "CHE1001",
    "CHE1018",
    "CHE1020"
}
```

```python
course_counts = {}

for i in range(1, len(lines)):
    entries = lines[i].split("\t")
    courseCode = entries[2]

    if courseCode in course_counts:
        course_counts[courseCode] += 1
    else:
        course_counts[courseCode] = 1
```

```python
filtered_lines = [lines[0]]
for i in range(1, len(lines)):
  entries = lines[i].split("\t")
  courseCode = entries[2]
  if courseCode in MOOC_COURSES:
    continue
  if course_counts.get(courseCode, 0) < 21:
    continue
  filtered_lines.append(lines[i])
lines = filtered_lines
print(len(lines))
```

2. Create a list of course codes registered for each roll number. Use a dictionary for representing the roll numbers. The key is the roll number, the value is the list of course codes associated with the roll number.

```python
registrations = dict()
for i in range(1, len(lines)):
  entries = lines[i].split("\t")
  rollNo = entries[0]
  courseCode = entries[2]
  if rollNo in registrations:
    registrations[rollNo].append(courseCode)
  else:
    registrations[rollNo] = [courseCode]
print(len(registrations))
print(registrations)
```

```python
from collections import Counter

def getDataFromRegistrations(registrations):
  ret = []
  for rollNo in registrations:
    ret.append(registrations[rollNo])
  return ret


def count_distinct_lists(list_of_lists):
    sorted_tuples = [tuple(sorted(lst)) for lst in list_of_lists]
    return Counter(sorted_tuples)
data = getDataFromRegistrations(registrations)

result = count_distinct_lists(data)
print(result)
```

```python
def min_groups(users):
    """
    users: dict(name -> set_of_items)
    returns: list of groups, each group is a list of user names
    """

    names = list(users.keys())
    conflict = {name: set() for name in names}

    for i in range(len(names)):
        for j in range(i + 1, len(names)):
            u, v = names[i], names[j]
            # If share common items, they conflict
            if set(users[u]) & set(users[v]):
                conflict[u].add(v)
                conflict[v].add(u)

    # Greedy coloring
    color = {}  # user -> group number
    for u in sorted(names, key=lambda x: len(conflict[x]), reverse=True):
        assigned_colors = {color[v] for v in conflict[u] if v in color}

        # choose smallest available color
        c = 0
        while c in assigned_colors:
            c += 1
        color[u] = c

    # Convert colors to group list
    groups = {}
    for user, c in color.items():
        groups.setdefault(c, []).append(user)

    return list(groups.values())
```

```python
groups = min_groups(users)
print(groups)
print(len(groups))
```

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
time_slots = [
    "9:00am-11:00am",
    "11:00am-1:00pm",
    # "1:00pm-2:00pm(Lunch Break)",
    "2:00pm-4:00pm"
]

def getslot(i):
    day = days[i // len(time_slots)]
    time = time_slots[i % len(time_slots)]
    return f"{day} {time}"

def assign_groups(num_groups=15):
    total_slots = len(days) * len(time_slots)
    if num_groups > total_slots:
        raise ValueError(f"Cannot schedule {num_groups} groups: only {total_slots} slots available (3 per day).")
    assignments = {}
    for i in range(num_groups):
        group_name = f"G{i+1}"
        assignments[group_name] = getslot(i)
    return assignments
if __name__ == "__main__":
    assignments = assign_groups(15)
    for g, slot in assignments.items():
        print(f"{g} -> {slot}")
```
